```python
def check_har_temperature_netcdf_flipping(netcdf_path, sample_date=None):
    """
    Check if HAR temperature NetCDF data is flipped with three simple plots.
    HAR uses curvilinear grid (2D lat/lon arrays) instead of 1D coordinate arrays.
    
    Parameters:
    -----------
    netcdf_path : str
        Path to the HAR temperature NetCDF file
    sample_date : str, optional
        Specific date to plot (format: 'YYYY-MM-DD'). If None, uses first available date.
    """
    import xarray as xr
    import matplotlib.pyplot as plt
    import numpy as np
    
    print(f"🌡️ Analyzing HAR NetCDF file: {netcdf_path}")
    
    try:
        # Load the NetCDF file
        ds = xr.open_dataset(netcdf_path)
        print(f"📊 Dataset info:")
        print(f"  Variables: {list(ds.variables.keys())}")
        print(f"  Dimensions: {dict(ds.dims)}")
        print(f"  Coordinates: {list(ds.coords.keys())}")
        
        # Find the temperature variable
        temp_var = None
        for var in ds.data_vars:
            if any(temp_name in var.lower() for temp_name in ['t2', 'temp', 'air']):
                temp_var = var
                break
        
        # Find the elevation variable
        elev_var = None
        for var in ds.data_vars:
            if any(elev_name in var.lower() for elev_name in ['elevation', 'elev', 'hgt']):
                elev_var = var
                break
        
        if temp_var is None:
            print(f"❌ Could not identify temperature variable from: {list(ds.data_vars)}")
            return None
        
        print(f"🌡️ Temperature variable: '{temp_var}'")
        if elev_var:
            print(f"🏔️ Elevation variable: '{elev_var}'")
        
        # === TEMPERATURE ANALYSIS ===
        print(f"\n{'='*60}")
        print(f"🌡️ TEMPERATURE ANALYSIS")
        print(f"{'='*60}")
        
        # Get temperature data
        temp_data = ds[temp_var]
        
        # Select a sample date for plotting
        if 'time' in temp_data.dims:
            if sample_date is not None:
                try:
                    temp_sample = temp_data.sel(time=sample_date, method='nearest')
                    actual_date = temp_sample.time.values
                    print(f"📅 Selected date: {sample_date} (actual: {actual_date})")
                except:
                    print(f"⚠️  Could not find date {sample_date}, using first available date")
                    temp_sample = temp_data.isel(time=0)
                    actual_date = temp_sample.time.values
                    print(f"📅 Using first available date: {actual_date}")
            else:
                temp_sample = temp_data.isel(time=0)
                actual_date = temp_sample.time.values
                print(f"📅 Using first available date: {actual_date}")
        else:
            temp_sample = temp_data
            actual_date = "No time dimension"
        
        # Get temperature values
        temp_values = temp_sample.values
        temp_units = temp_sample.attrs.get('units', 'unknown')
        print(f"🌡️ Temperature units: {temp_units}")
        print(f"🌡️ Temperature range: {np.nanmin(temp_values):.2f} to {np.nanmax(temp_values):.2f}")
        print(f"🌡️ Temperature shape: {temp_values.shape}")
        
        # Convert Kelvin to Celsius if needed
        if temp_units.lower() in ['k', 'kelvin'] or np.nanmin(temp_values) > 200:
            temp_values = temp_values - 273.15
            temp_units = '°C'
            print(f"🔄 Converted from Kelvin to Celsius")
            print(f"🌡️ Temperature range (°C): {np.nanmin(temp_values):.2f} to {np.nanmax(temp_values):.2f}")
        
        # === COORDINATE ANALYSIS ===
        print(f"\n{'='*60}")
        print(f"📍 COORDINATE ANALYSIS")
        print(f"{'='*60}")
        
        # HAR has 2D lat/lon arrays
        if 'lat' in ds.coords and 'lon' in ds.coords:
            lat_2d = ds.coords['lat'].values
            lon_2d = ds.coords['lon'].values
            
            print(f"🗺️  2D Latitude shape: {lat_2d.shape}")
            print(f"🗺️  2D Longitude shape: {lon_2d.shape}")
            print(f"🗺️  Latitude range: {lat_2d.min():.4f} to {lat_2d.max():.4f}")
            print(f"🗺️  Longitude range: {lon_2d.min():.4f} to {lon_2d.max():.4f}")
            
            # Check latitude orientation (compare first and last rows)
            first_row_lat = lat_2d[0, :].mean()
            last_row_lat = lat_2d[-1, :].mean()
            
            print(f"🗺️  First row (j=0) mean latitude: {first_row_lat:.4f}")
            print(f"🗺️  Last row (j=-1) mean latitude: {last_row_lat:.4f}")
            
            if first_row_lat > last_row_lat:
                print(f"  ⚠️  Latitude DECREASES from first to last row (north to south)")
                print(f"     Data appears to be oriented with NORTH at TOP (j=0)")
                needs_flip = True
            else:
                print(f"  ✅ Latitude INCREASES from first to last row (south to north)")
                print(f"     Data appears to be oriented with SOUTH at TOP (j=0)")
                needs_flip = False
            
            # Check some corner coordinates
            print(f"\n📍 Corner Coordinates:")
            print(f"  Top-left [0,0]:     Lat={lat_2d[0, 0]:.4f}, Lon={lon_2d[0, 0]:.4f}")
            print(f"  Top-right [0,-1]:   Lat={lat_2d[0, -1]:.4f}, Lon={lon_2d[0, -1]:.4f}")
            print(f"  Bottom-left [-1,0]: Lat={lat_2d[-1, 0]:.4f}, Lon={lon_2d[-1, 0]:.4f}")
            print(f"  Bottom-right [-1,-1]: Lat={lat_2d[-1, -1]:.4f}, Lon={lon_2d[-1, -1]:.4f}")
            
        else:
            print(f"❌ No lat/lon coordinates found")
            lat_2d = None
            lon_2d = None
            needs_flip = None
        
        # === ELEVATION ANALYSIS ===
        elev_analysis = {}
        if elev_var:
            print(f"\n{'='*60}")
            print(f"🏔️ ELEVATION ANALYSIS")
            print(f"{'='*60}")
            
            elev_data = ds[elev_var]
            elev_values = elev_data.values
            elev_units = elev_data.attrs.get('units', 'unknown')
            
            print(f"🏔️ Elevation units: {elev_units}")
            print(f"🏔️ Elevation range: {np.nanmin(elev_values):.1f} to {np.nanmax(elev_values):.1f}")
            print(f"🏔️ Elevation shape: {elev_values.shape}")
            print(f"🏔️ Mean elevation: {np.nanmean(elev_values):.1f} {elev_units}")
            
            elev_analysis = {
                'values': elev_values,
                'units': elev_units
            }
        
        # === PLOTTING ===
        print(f"\n{'='*60}")
        print(f"📊 CREATING COMPARISON PLOTS")
        print(f"{'='*60}")
        
        # Determine number of subplots
        n_vars = 1 + (1 if elev_var else 0)
        
        if n_vars == 2:
            fig, axes = plt.subplots(2, 3, figsize=(20, 12))
            temp_axes = axes[0, :]
            elev_axes = axes[1, :]
        else:
            fig, axes = plt.subplots(1, 3, figsize=(20, 6))
            temp_axes = axes
            elev_axes = None
        
        # === TEMPERATURE PLOTS ===
        # Plot 1: With lat/lon coordinates (pcolormesh for 2D coords)
        ax1 = temp_axes[0]
        if lat_2d is not None and lon_2d is not None:
            im1 = ax1.pcolormesh(lon_2d, lat_2d, temp_values, cmap='RdBu_r', shading='auto')
            plt.colorbar(im1, ax=ax1, label=f'Temperature ({temp_units})')
            ax1.set_title('Temperature: 2D Lat/Lon Coordinates\n(pcolormesh with curvilinear grid)')
            ax1.set_xlabel('Longitude')
            ax1.set_ylabel('Latitude')
        else:
            ax1.text(0.5, 0.5, 'No lat/lon coordinates available', 
                    transform=ax1.transAxes, ha='center', va='center', fontsize=14)
            ax1.set_title('Temperature: No Coordinates Available')
        
        # Plot 2: With cell indices (original data orientation)
        ax2 = temp_axes[1]
        im2 = ax2.imshow(temp_values, cmap='RdBu_r', aspect='auto', origin='upper')
        plt.colorbar(im2, ax=ax2, label=f'Temperature ({temp_units})')
        ax2.set_title('Temperature: Cell Indices (Original)\nY-axis: j=0 at top, increasing downward')
        ax2.set_xlabel('Column Index (i, west_east)')
        ax2.set_ylabel('Row Index (j, south_north)')
        
        # Add corner values
        if temp_values.shape[0] > 1 and temp_values.shape[1] > 1:
            ax2.text(0.02, 0.98, f'Top-left [0,0]: {temp_values[0,0]:.1f}°C', 
                    transform=ax2.transAxes, verticalalignment='top',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))
            ax2.text(0.98, 0.02, f'Bottom-right [-1,-1]: {temp_values[-1,-1]:.1f}°C', 
                    transform=ax2.transAxes, verticalalignment='bottom', horizontalalignment='right',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))
        
        # Plot 3: With cell indices flipped (j=0 at bottom)
        ax3 = temp_axes[2]
        im3 = ax3.imshow(temp_values, cmap='RdBu_r', aspect='auto', origin='lower')
        plt.colorbar(im3, ax=ax3, label=f'Temperature ({temp_units})')
        ax3.set_title('Temperature: Cell Indices (Flipped)\nY-axis: j=0 at bottom, increasing upward')
        ax3.set_xlabel('Column Index (i, west_east)')
        ax3.set_ylabel('Row Index (j, south_north)')
        
        # Add corner values for flipped version
        if temp_values.shape[0] > 1 and temp_values.shape[1] > 1:
            ax3.text(0.02, 0.02, f'Bottom-left (array[0,0]): {temp_values[0,0]:.1f}°C', 
                    transform=ax3.transAxes, verticalalignment='bottom',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))
            ax3.text(0.98, 0.98, f'Top-right (array[-1,-1]): {temp_values[-1,-1]:.1f}°C', 
                    transform=ax3.transAxes, verticalalignment='top', horizontalalignment='right',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))
        
        # === ELEVATION PLOTS (if available) ===
        if elev_analysis and elev_axes is not None:
            elev_values = elev_analysis['values']
            elev_units = elev_analysis['units']
            
            # Plot 1: With lat/lon coordinates
            ax1 = elev_axes[0]
            if lat_2d is not None and lon_2d is not None:
                im1 = ax1.pcolormesh(lon_2d, lat_2d, elev_values, cmap='terrain', shading='auto')
                plt.colorbar(im1, ax=ax1, label=f'Elevation ({elev_units})')
                ax1.set_title('Elevation: 2D Lat/Lon Coordinates')
                ax1.set_xlabel('Longitude')
                ax1.set_ylabel('Latitude')
            
            # Plot 2: Original orientation
            ax2 = elev_axes[1]
            im2 = ax2.imshow(elev_values, cmap='terrain', aspect='auto', origin='upper')
            plt.colorbar(im2, ax=ax2, label=f'Elevation ({elev_units})')
            ax2.set_title('Elevation: Cell Indices (Original)')
            ax2.set_xlabel('Column Index (i)')
            ax2.set_ylabel('Row Index (j)')
            
            # Plot 3: Flipped orientation
            ax3 = elev_axes[2]
            im3 = ax3.imshow(elev_values, cmap='terrain', aspect='auto', origin='lower')
            plt.colorbar(im3, ax=ax3, label=f'Elevation ({elev_units})')
            ax3.set_title('Elevation: Cell Indices (Flipped)')
            ax3.set_xlabel('Column Index (i)')
            ax3.set_ylabel('Row Index (j)')
        
        plt.tight_layout()
        plt.show()
        
        # === ANALYSIS SUMMARY ===
        print(f"\n{'='*60}")
        print(f"📊 HAR DATA ANALYSIS SUMMARY")
        print(f"{'='*60}")
        
        if lat_2d is not None:
            print(f"\n  Coordinate System:")
            print(f"    Grid type: CURVILINEAR (2D lat/lon arrays)")
            print(f"    Projection: Lambert Conformal Conic (HAR v2)")
            print(f"    Latitude range: {lat_2d.min():.4f} to {lat_2d.max():.4f}")
            print(f"    Longitude range: {lon_2d.min():.4f} to {lon_2d.max():.4f}")
            
            if needs_flip:
                print(f"\n  ⚠️  DATA ORIENTATION:")
                print(f"      First row (j=0) is at HIGHER latitude than last row (j=-1)")
                print(f"      This means j=0 is in the NORTH")
                print(f"      Standard orientation would have j=0 in SOUTH")
                print(f"      🤔 Check if this matches your expected orientation")
            else:
                print(f"\n  ✅ DATA ORIENTATION:")
                print(f"      First row (j=0) is at LOWER latitude than last row (j=-1)")
                print(f"      This means j=0 is in the SOUTH")
                print(f"      This is standard orientation")
        
        print(f"\n  Temperature Statistics:")
        print(f"    Shape: {temp_values.shape} (rows, columns)")
        print(f"    Range: {np.nanmin(temp_values):.2f} to {np.nanmax(temp_values):.2f} {temp_units}")
        print(f"    Mean: {np.nanmean(temp_values):.2f} {temp_units}")
        
        if elev_analysis:
            print(f"\n  Elevation Statistics:")
            print(f"    Range: {np.nanmin(elev_values):.1f} to {np.nanmax(elev_values):.1f} {elev_units}")
            print(f"    Mean: {np.nanmean(elev_values):.1f} {elev_units}")
        
        print(f"\n💡 HAR-Specific Notes:")
        print(f"  - HAR uses Lambert Conformal Conic projection")
        print(f"  - Coordinates are CURVILINEAR (2D arrays, not regular grid)")
        print(f"  - Each grid cell has unique lat/lon coordinates")
        print(f"  - Latitude typically INCREASES from south to north")
        print(f"  - If your plots show mountains in wrong location, orientation may be flipped")
        
        if needs_flip is not None:
            print(f"\n  Current Orientation Status:")
            if needs_flip:
                print(f"    ⚠️  Data appears NORTH-UP (j=0 at north)")
                print(f"    🔧 You may need to flip the j-axis if expecting south-up")
            else:
                print(f"    ✅ Data appears SOUTH-UP (j=0 at south) - STANDARD")
        
        # Return results
        results = {
            'coordinates': {'lat': lat_2d, 'lon': lon_2d},
            'needs_flip': needs_flip,
            'temperature': {
                'values': temp_values,
                'units': temp_units
            },
            'dataset': ds
        }
        
        if elev_analysis:
            results['elevation'] = elev_analysis
        
        return results
        
    except Exception as e:
        print(f"❌ Error analyzing HAR NetCDF file: {e}")
        import traceback
        traceback.print_exc()
        return None
```
